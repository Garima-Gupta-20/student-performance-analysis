# Student Performance Analysis Project

This project analyzes student academic performance using Python.
The analysis focuses on marks, attendance, and subject-wise performance.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df = pd.read_excel(r'C:\\Users\\pc\\Desktop\\Student performence analysis\\Data\\Student_Performance_Dataset.xlsx')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
subject_avg = df.groupby('Subject')['Total_Marks'].mean()
subject_avg
```

```python
plt.figure()
subject_avg.plot(kind='bar', color='pink',edgecolor='black')
plt.title('Average Total Marks by Subject')
plt.xlabel('Subject')
plt.ylabel('Average Total Marks')
plt.grid(False)
plt.show()
```

## Subject-wise Performance Insight
- Computer Science has the highest average marks.
- Mathematics shows comparatively lower performance.


```python
plt.figure()
sns.scatterplot(x = "Attendance_Percentage",y = "Total_Marks",data = df,color = 'purple',edgecolor='black',marker = "D",)
plt.title('Attendance Percentage vs Total Marks')
plt.xlabel('Attendance Percentage')
plt.ylabel('Total Marks')
plt.grid(True)
plt.show()
```

```python
df["Attendance_Percentage"].corr(df["Total_Marks"])
```

## Attendance vs Performance Insight
- Students with higher attendance generally score higher marks.
- Attendance shows a positive correlation with total marks.
## Key Insight
The correlation coefficient between attendance and total marks is 0.97, which indicates a very strong positive relationship. Students with higher attendance consistently achieve better academic results.


```python
plt.figure()
sns.boxplot(x = "Subject",y = "Total_Marks",data = df,palette = "Set2")
plt.title('Total Marks Distribution by Subject')
plt.xlabel('Subject')
plt.ylabel('Total Marks')
plt.grid(True)
plt.show()
```

```python
top_students = df.sort_values(by='Total_Marks', ascending=False).head(10)
top_students
```

```python
weak_students = df[df["Total_Marks"] < 60]
weak_students
```

## Conclusion
- Attendance plays a critical role in academic success.
- Computer Science students performed best on average.
- Students with low attendance require early academic intervention.
